### Understanding the characteristics of the `is_fraud` == 1

```python
df[df['is_fraud'] == 1]
```

Fraud detection in transactions is the process of identifying transactions that are intentionally deceptive or unauthorized (e.g., stolen card use, account takeover, money‑laundering attempts) using rules, statistical methods, and machine learning.

Key signals/features

Transaction amount, time (time_ind), merchant/recipient (dst_acc), sender (src_acc)
Geolocation / device / IP / channel
Historical behavior / velocity features (frequency, spend patterns, sudden changes)
Derived features: time since last txn, rolling means, count of recipients, categorical encodings

Labeling approaches

Supervised: labelled examples (is_fraud) from chargebacks/alerts/investigations
Semi/unsupervised: anomaly detection/clustering when labels are scarce
Common methods

Rule-based heuristics (thresholds, blacklists)
Feature engineering + classical ML (logistic regression, tree ensembles)
Deep learning / sequence models for behavioral patterns
Hybrid systems: fast rules for real-time blocking + ML for scoring/alerting

Evaluation metrics & considerations

Precision, recall, F1, AUPRC (precision-recall focus because of class imbalance)
False positives cost (customer friction) vs false negatives (loss)
Use time-based cross-validation to avoid leakage and mimic production

Operational challenges

Severe class imbalance — use sampling, class weights, anomaly methods
Concept drift — regular retraining and monitoring
Latency & explainability requirements for real-time decisions
Feedback loop: label delay and label quality (chargebacks are delayed/noisy)
